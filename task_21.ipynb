```python
import pandas as pd
from typing import List
from transformers import pipeline
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score


def load_data(neg_path: str, pos_path: str) -> pd.DataFrame:
    with open(neg_path, "r", encoding='utf-8', errors='ignore') as f:
        texts_neg = f.read().splitlines()

    with open(pos_path, "r", encoding='utf-8', errors='ignore') as f:
        texts_pos = f.read().splitlines()

    df_neg = pd.DataFrame({'text': texts_neg, 'label': 0})
    df_pos = pd.DataFrame({'text': texts_pos, 'label': 1})

    return pd.concat([df_neg, df_pos], ignore_index=True)


def evaluate_model(y_true: pd.Series, y_pred: List[int], model_name: str) -> None:
    print(f"\n{'='*50}")
    print(f"Модель: {model_name}")
    print(f"{'='*50}")
    print(f"Accuracy:  {accuracy_score(y_true, y_pred):.4f}")
    print(f"Precision: {precision_score(y_true, y_pred):.4f}")
    print(f"Recall:    {recall_score(y_true, y_pred):.4f}")
    print(f"F1-score:  {f1_score(y_true, y_pred):.4f}")
    print(f"{'='*50}\n")


def main():
    df = load_data('rt-polarity.neg', 'rt-polarity.pos')

    X_train, X_test, y_train, y_test = train_test_split(
        df['text'],
        df['label'],
        test_size=0.2,
        random_state=42,
        stratify=df['label']
    )

    vectorizer = TfidfVectorizer(max_features=5000)
    X_train_tfidf = vectorizer.fit_transform(X_train)
    X_test_tfidf = vectorizer.transform(X_test)

    lr_model = LogisticRegression(max_iter=1000)
    lr_model.fit(X_train_tfidf, y_train)
    lr_preds = lr_model.predict(X_test_tfidf)
    evaluate_model(y_test, lr_preds, "Logistic Regression (TF-IDF)")

    svm_model = LinearSVC(dual=False)
    svm_model.fit(X_train_tfidf, y_train)
    svm_preds = svm_model.predict(X_test_tfidf)
    evaluate_model(y_test, svm_preds, "Support Vector Machine (TF-IDF)")

    sentiment_pipeline = pipeline(
        "sentiment-analysis",
        model="distilbert-base-uncased-finetuned-sst-2-english",
        truncation=True
    )

    test_texts = X_test.tolist()
    transformer_results = sentiment_pipeline(test_texts)
    transformer_preds = [
        1 if res['label'] == 'POSITIVE' else 0
        for res in transformer_results
    ]

    evaluate_model(y_test, transformer_preds, "Modern Transformer (DistilBERT)")


if __name__ == "__main__":
    main()
```
