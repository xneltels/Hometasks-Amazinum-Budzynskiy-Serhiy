```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.stattools import adfuller
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

warnings.filterwarnings("ignore")
plt.style.use('seaborn-v0_8-darkgrid')
```

```python
def load_and_prepare_data(filepath):
    df = pd.read_csv(filepath)

    if 'Datetime' not in df.columns:
        possible_date_cols = [col for col in df.columns if 'date' in col.lower() or 'time' in col.lower()]
        if possible_date_cols:
            df.rename(columns={possible_date_cols[0]: 'Datetime'}, inplace=True)
        else:
            df.rename(columns={df.columns[0]: 'Datetime'}, inplace=True)

    df['Datetime'] = pd.to_datetime(df['Datetime'])
    df.set_index('Datetime', inplace=True)

    ts = df['Consumption'].ffill()
    return ts.resample('D').sum()

filepath = r"C:\Users\Admin\Downloads\NetCrackerPro4.1\electricity_consumption_and_production.csv"
ts = load_and_prepare_data(filepath)
```

```python
plt.figure(figsize=(16, 5))
plt.plot(ts, color='teal', linewidth=1.5)
plt.title('Щоденне споживання електроенергії (Загальний вигляд)', fontsize=14)
plt.xlabel('Дата')
plt.ylabel('Споживання')
plt.tight_layout()
plt.show()

print("--- Розширений тест Дікі-Фуллера (ADF Test) ---")
dickey_fuller = adfuller(ts)
print(f"ADF Statistic: {dickey_fuller[0]:.4f}")
print(f"p-value: {dickey_fuller[1]:.4f}")
if dickey_fuller[1] < 0.05:
    print("Висновок: Часовий ряд є стаціонарним (відкидаємо нульову гіпотезу).")
else:
    print("Висновок: Часовий ряд є нестаціонарним (не можемо відкинути нульову гіпотезу).")
```

```python
def calculate_mape(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    non_zero_idx = y_true != 0
    if len(y_true[non_zero_idx]) == 0:
        return 0.0
    return np.mean(np.abs((y_true[non_zero_idx] - y_pred[non_zero_idx]) / y_true[non_zero_idx])) * 100

def evaluate_forecast(y_true, y_pred, horizon, model_name):
    y_t = y_true[:horizon]
    y_p = y_pred[:horizon]

    mae = mean_absolute_error(y_t, y_p)
    rmse = np.sqrt(mean_squared_error(y_t, y_p))
    mape = calculate_mape(y_t, y_p)

    print(f"{model_name:<18} | Horizon: {horizon:2} days -> MAE: {mae:8.2f} | RMSE: {rmse:8.2f} | MAPE: {mape:6.2f}%")
    return {'Model': model_name, 'Horizon': horizon, 'MAE': mae, 'RMSE': rmse, 'MAPE': mape}

def prepare_ml_features(ts, n_lags=7):
    df = pd.DataFrame({'y': ts})
    for i in range(1, n_lags + 1):
        df[f'lag_{i}'] = df['y'].shift(i)
    df = df.dropna()
    return df.drop('y', axis=1), df['y']

def forecast_ml_model(train_ts, steps, model, n_lags=7):
    X_train, y_train = prepare_ml_features(train_ts, n_lags)
    model.fit(X_train, y_train)
    current_features = train_ts[-n_lags:].values[::-1]
    predictions = []
    for _ in range(steps):
        pred = model.predict(current_features.reshape(1, -1))[0]
        predictions.append(pred)
        current_features = np.insert(current_features, 0, pred)[:-1]
    return predictions
```

```python
test_size = 45
train = ts.iloc[:-test_size]
test = ts.iloc[-test_size:]

print("Навчання SARIMA...")
sarima_model = SARIMAX(train, order=(1, 1, 1), seasonal_order=(1, 1, 1, 7)).fit(disp=False)
sarima_preds = sarima_model.forecast(steps=test_size)

print("Навчання Random Forest...")
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_preds = pd.Series(forecast_ml_model(train, test_size, rf_model), index=test.index)

print("Навчання Gradient Boosting...")
gb_model = GradientBoostingRegressor(n_estimators=100, random_state=42)
gb_preds = pd.Series(forecast_ml_model(train, test_size, gb_model), index=test.index)
print("Навчання завершено!")
```

```python
results = []
print("="*75)
for horizon in [15, 30, 45]:
    results.append(evaluate_forecast(test, sarima_preds, horizon, "SARIMA"))
    results.append(evaluate_forecast(test, rf_preds, horizon, "Random Forest"))
    results.append(evaluate_forecast(test, gb_preds, horizon, "Gradient Boosting"))
    print("-" * 75)
```

```python
fig, axes = plt.subplots(3, 1, figsize=(16, 15), sharex=True)
plot_train_days = 60

models_data = [
    ("SARIMA", sarima_preds, 'green'),
    ("Random Forest", rf_preds, 'red'),
    ("Gradient Boosting", gb_preds, 'orange')
]

for i, (name, preds, color) in enumerate(models_data):
    axes[i].plot(train.index[-plot_train_days:], train.iloc[-plot_train_days:], label='Train', color='gray', alpha=0.5)
    axes[i].plot(test.index, test, label='Actual Test', color='blue', linewidth=2)

    axes[i].plot(preds.index, preds, label=f'{name} Forecast', color=color, linestyle='--', linewidth=2)

    axes[i].axvline(x=test.index[0], color='black', linestyle=':', alpha=0.7)

    axes[i].set_title(f'Прогноз моделі: {name}', fontsize=12, fontweight='bold')
    axes[i].legend(loc='upper left')
    axes[i].grid(True, alpha=0.3)

plt.xlabel('Дата', fontsize=12)
plt.tight_layout()
plt.show()
```
